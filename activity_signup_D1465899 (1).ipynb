本作業整理「桌遊體驗」與「電影欣賞」兩場活動的原始報名文字，清理空白與大小寫後轉為串列保留原始順序，接著以集合去重並進行交集、聯集與差集比較，排序後取前兩位作為通知名單預覽，最後透過字典彙整人數與各場應收金額並輸出完整摘要。

## 測試一：原始資料

```python
# 測試一：原始資料設定（保留原始文字變數）
activity_names = ("桌遊體驗", "電影欣賞")  # tuple：保存固定的兩個活動名稱
board_raw = " Amy, Ben, amy, Cora "       # str：桌遊原始報名文字
movie_raw = " ben, Dora, ERIC, ben "      # str：電影原始報名文字
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}  # dict：各活動每人費用

# 讀取活動名稱並整理成保留重複的串列（移除半形空白、轉小寫、依逗號拆分）
board_event, movie_event = activity_names
board_list = board_raw.replace(" ", "").lower().split(",")
movie_list = movie_raw.replace(" ", "").lower().split(",")

# 轉成集合去重並進行集合運算
board_set = set(board_list)
movie_set = set(movie_list)

both_set = board_set & movie_set          # 交集：共同參加者
all_set = board_set | movie_set           # 聯集：全部參加者（至少參加一場）
only_board_set = board_set - movie_set    # 差集：只參加桌遊者
only_movie_set = movie_set - board_set    # 差集：只參加電影者

# 排序成串列以利顯示與切片
board_sorted = sorted(board_set)
movie_sorted = sorted(movie_set)
both_sorted = sorted(both_set)
only_board_sorted = sorted(only_board_set)
only_movie_sorted = sorted(only_movie_set)
all_sorted = sorted(all_set)

# 通知名單預覽：全部參加者排序後取前兩位
preview_notify = all_sorted[:2]

# 計算人數與應收金額（保留數值型態於 summary 字典）
board_count = len(board_set)
movie_count = len(movie_set)
both_count = len(both_set)
total_unique_count = len(all_set)
total_visits = board_count + movie_count

board_revenue = board_count * unit_fees[board_event]
movie_revenue = movie_count * unit_fees[movie_event]
total_revenue = board_revenue + movie_revenue

summary = {
    "board_count": board_count,
    "movie_count": movie_count,
    "both_count": both_count,
    "total_unique_count": total_unique_count,
    "total_visits": total_visits,
    "board_revenue": board_revenue,
    "movie_revenue": movie_revenue,
    "total_revenue": total_revenue,
}

# 輸出完整摘要
print("HW09 活動報名摘要")
print(f"活動：{board_event}，每人 {unit_fees[board_event]:.2f} 元")
print(f"活動：{movie_event}，每人 {unit_fees[movie_event]:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{board_sorted}")
print(f"電影去重名單：{movie_sorted}")
print(f"桌遊人數：{summary['board_count']} 人")
print(f"電影人數：{summary['movie_count']} 人")
print(f"共同參加者：{both_sorted}")
print(f"共同人數：{summary['both_count']} 人")
print(f"只參加桌遊者：{only_board_sorted}")
print(f"只參加電影者：{only_movie_sorted}")
print(f"全部參加者：{all_sorted}")
print(f"總不重複人數：{summary['total_unique_count']} 人")
print(f"報名人次：{summary['total_visits']} 人次")
print(f"通知名單前兩位：{preview_notify}")
print(f"桌遊應收：{summary['board_revenue']:.2f} 元")
print(f"電影應收：{summary['movie_revenue']:.2f} 元")
print(f"總應收：{summary['total_revenue']:.2f} 元")
```
